Inilah dasar dari bagaimana objek-objek dalam program bisa "berkomunikasi" dan saling memengaruhi satu sama lain, layaknya karakter-karakter dalam game yang benar-benar berinteraksi, bukan cuma nama-nama di daftar.

```python
# latihan 4
class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        # Enkapsulasi: HP bersifat Private (hanya bisa diakses di dalam class ini)
        self.__hp = hp_awal

    # GETTER: Cara resmi melihat nilai HP yang bersifat private
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0  # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        # Kita pakai setter/getter bahkan di dalam class sendiri agar aman
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")


# -- Uji Coba --
hero1 = Hero("Layla", 100)

# hero1.__hp = 9999   # GAGAL (Tidak akan mengubah hp asli)
# print(hero1.__hp)   # ERROR (Tidak bisa dibaca langsung)

hero1.set_hp(-50)       # Coba set negatif
print(hero1.get_hp())   # Output: 0 (Karena dicegat oleh logika Setter)

print(f"Mencoba akses paksa: {hero1._Hero__hp}")
```

Analisis 4: percobaan hacking jawaban: Nilai HP tetap muncul saat mengakses hero1._Hero__hp. Hal ini karena Python menggunakan Name Mangling, yaitu mengubah nama atribut private menjadi _Hero__hp. Meskipun masih bisa diakses dengan cara tersebut, tindakan ini tidak disarankan karena melanggar prinsip Encapsulation dan dapat merusak keamanan data.

Uji Validasi Jawaban: Jika logika if dan elif pada set_hp() dihapus, lalu menjalankan hero1.set_hp(-100), maka HP Hero akan menjadi -100. Data tersebut menjadi tidak valid karena HP seharusnya tidak boleh bernilai negatif. Oleh karena itu, method Setter sangat penting untuk memvalidasi data sebelum disimpan sehingga integritas data dalam game tetap terjaga

```python
# latihan 5
from abc import ABC, abstractmethod

# 1. Interface / Abstract Class
# Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.
class GameUnit(ABC):
    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass


# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    # Kita WAJIB membuat method serang, kalau tidak akan Error
    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")


class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis

    # Implementasi serang versi Monster
    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")


# -- Uji Coba --
# unit = GameUnit()   # ERROR! Abstract class tidak bisa jadi objek.

h = Hero("Alucard")
m = Monster("Serigala")

h.info()
m.info()
```

Menggelar Kontrak Jawaban: Jika method serang() pada class Hero dihapus, program akan menghasilkan error:
TypeError: Can't instantiate abstract class Hero with abstract method seran

rti dari error tersebut adalah class Hero belum memenuhi aturan (kontrak) yang dibuat oleh abstract class GameUnit, yaitu harus memiliki method serang(). Karena method tersebut belum dibuat, Python tidak mengizinkan objek Hero dibuat.

Konsekuensinya, jika kita lupa membuat method yang sudah dijanjikan di Interface, class tersebut tidak dapat diinstansiasi (dibuat objek) sehingga program tidak dapat dijalankan sebagaimana mestinya.

Mencetak Cetakan Jawaban: Class GameUnit tidak boleh dibuat menjadi objek karena merupakan abstract class yang hanya berfungsi sebagai kerangka dasar (blueprint). Abstract class digunakan untuk menentukan method yang wajib dimiliki oleh semua class turunannya, seperti Hero dan Monster.
Walaupun tidak bisa dibuat menjadi objek, GameUnit tetap berguna sebagai pedoman agar semua class turunannya memiliki struktur dan method yang sama, sehingga program menjadi lebih rapi, konsisten, dan mudah dikembangkan.

```python
# latihan 6
# Parent Class
class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")


# Child Class 1
class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")


# Child Class 2
class Archer(Hero):
    def tembak_panah(self):
        print(f"{self.nama} memanah dari jauh! Jleb!")


# Child Class 3
class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")

class Healer(Hero):
    def serang(self):
        print(f"{self.nama} tidak menyerang, tapi menyembuhkan teman!")

# Penerapan Polymorphism
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord"),
    Healer("Rafaela")

]

print("--- PERANG DIMULAI ---")

for pahlawan in pasukan:
    pahlawan.serang()
```

Uji Skalabilitas:

, program tetap berjalan dengan lancar tanpa mengubah kode pada perulangan (for pahlawan in pasukan:). Class Healer dapat langsung digunakan karena memiliki method serang() seperti class lainnya.

keuntungan Polimorfisme memudahkan programmer menambahkan karakter baru tanpa mengubah kode yang sudah ada. Cukup membuat class baru dengan method yang sama (serang()), maka karakter tersebut dapat langsung digunakan dalam program.

Konsistense Penamaan

Program akan menghasilkan error karena saat perulangan dijalankan, program memanggil pahlawan.serang(), sedangkan class Archer sudah tidak memiliki method serang().

pada konsep Polymorphism, semua class turunan harus memiliki nama method yang sama agar dapat dipanggil dengan satu perintah yang sama. Jika nama method berbeda, objek tersebut tidak dapat menjalankan perintah yang diberikan sehingga menyebabkan error.
